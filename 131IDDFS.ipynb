```python
def depth_limited_dfs(start, goal, depth_limit):
    """Performs DFS up to a specified maximum depth limit."""
    # Stack stores: (state, path_taken, current_depth)
    stack = [(start, [start], 0)]
    
    # Map state to the shallowest depth at which it was visited
    visited = {start: 0}

    while stack:
        state, path, depth = stack.pop()

        if state == goal:
            return path

        # If limit reached, do not expand neighbors further
        if depth >= depth_limit:
            continue

        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = [
            (-1, 0), # Up
            (1, 0),  # Down
            (0, -1), # Left
            (0, 1)   # Right
        ]

        for dr, dc in moves:
            r, c = row + dr, col + dc

            if 0 <= r < 3 and 0 <= c < 3:
                new_pos = r * 3 + c
                new_state = list(state)
                new_state[zero], new_state[new_pos] = (
                    new_state[new_pos], new_state[zero]
                )
                new_state = tuple(new_state)

                next_depth = depth + 1

                # Visit state if not seen yet OR if found via a shorter path in current iteration
                if new_state not in visited or next_depth < visited[new_state]:
                    visited[new_state] = next_depth
                    stack.append((new_state, path + [new_state], next_depth))

    return None


def id_dfs(start, goal, max_depth=50):
    """Iterative Deepening DFS: Progressively increases depth limit."""
    for limit in range(max_depth + 1):
        result = depth_limited_dfs(start, goal, limit)
        if result is not None:
            print(f"Optimal solution found at depth limit: {limit}")
            return result
    return None


# Initial and Goal States
start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

# Execute IDDFS
result = id_dfs(start, goal)

if result:
    print("Solution Found!")
    print("Number of Moves:", len(result) - 1)

    for step, state in enumerate(result):
        print("\nStep", step)
        for i in range(0, 9, 3):
            print(state[i:i+3])
else:
    print("No solution found within max depth")
```
