```python
import random
import math

N = 8

def conflicts(board):
    count = 0

    for i in range(N):
        for j in range(i + 1, N):
            if board[i] == board[j]:
                count += 1
            elif abs(board[i] - board[j]) == abs(i - j):
                count += 1

    return count


def simulated_annealing():
    for restart in range(100):
        board = [random.randint(0, 7) for _ in range(N)]
        temperature = 10.0

        for step in range(5000):
            current_cost = conflicts(board)

            if current_cost == 0:
                return board

            new_board = board[:]
            col = random.randint(0, 7)
            new_row = random.choice(
                [r for r in range(N) if r != board[col]]
            )
            new_board[col] = new_row

            new_cost = conflicts(new_board)
            delta = new_cost - current_cost

            if delta <= 0 or random.random() < math.exp(
                -delta / temperature
            ):
                board = new_board

            temperature *= 0.995

    return None


solution = simulated_annealing()

if solution is not None:
    print("Solution found:")
    for row in range(N):
        for col in range(N):
            if solution[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()

    print("Number of conflicts:", conflicts(solution))
else:
    print("No solution found. Try running again.")
```
