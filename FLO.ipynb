```python
def unify(a, b, subs):
    if a == b:
        return subs

    if isinstance(a, str) and a.islower() and len(a) == 1:
        subs[a] = b
        return subs

    if isinstance(b, str) and b.islower() and len(b) == 1:
        subs[b] = a
        return subs

    if isinstance(a, tuple) and isinstance(b, tuple):
        if len(a) != len(b) or a[0] != b[0]:
            return None

        for x, y in zip(a[1:], b[1:]):
            subs = unify(x, y, subs)
            if subs is None:
                return None
        return subs

    return None


tests = [
    (("Eats", "x", "Apple"), ("Eats", "Riya", "y")),
    (("Loves", "x", "Jatin"), ("Loves", "Khushi", "y")),
    (("Knows", "John", "x"), ("Knows", "y", "Elisabeth")),
    (("P", "x", "x"), ("P", "a", "b")),
    (("Loves", "x"), ("Likes", "y"))
]

print("\n          UNIFICATION ALGORITHM")
print("  " + "─" * 38)

for i, (a, b) in enumerate(tests, 1):
    result = unify(a, b, {})

    print(f"\n  TEST CASE {i}")
    print(f"  Expression 1 : {a}")
    print(f"  Expression 2 : {b}")

    if result is None:
        print("  Result       : FAILED")
    else:
        print(f"  Substitution : {result}")
        print("  Result       : SUCCESS")

print("\n  " + "─" * 38)
```
