```python
board = [" "] * 9
player = "X"

win = [
    (0, 1, 2), (0, 3, 6), (1, 4, 7), (2, 5, 8),
    (3, 4, 5), (6, 7, 8), (0, 4, 8), (2, 4, 6)
]

for i in range(9):
    print(board[0:3])
    print(board[3:6])
    print(board[6:9])

    pos = int(input(player + " enter position (1-9): ")) - 1

    if board[pos] == " ":
        board[pos] = player
    else:
        print("Position occupied")
        continue

    if any(board[a] == board[b] == board[c] != " " for a, b, c in win):
        print(board[0:3])
        print(board[3:6])
        print(board[6:9])
        print(player, "wins!")
        break

    player = "O" if player == "X" else "X"

else:
    print("Draw")
```
