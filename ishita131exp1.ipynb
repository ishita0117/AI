```python

board = [" " for _ in range(9)]

def show_board():
    print("\n")
    print(" " + board[0] + " | " + board[1] + " | " + board[2])
    print("---+---+---")
    print(" " + board[3] + " | " + board[4] + " | " + board[5])
    print("---+---+---")
    print(" " + board[6] + " | " + board[7] + " | " + board[8])
    print()


def check_winner(player):
    win = [
        [0,1,2], [3,4,5], [6,7,8],
        [0,3,6], [1,4,7], [2,5,8],
        [0,4,8], [2,4,6]
    ]

    for a, b, c in win:
        if board[a] == board[b] == board[c] == player:
            return True
    return False


def play_game():

    global board
    board = [" " for _ in range(9)]
    player = "X"

    print("\nTIC-TAC-TOE")
    print("Positions are:")
    print(" 1 | 2 | 3")
    print("---+---+---")
    print(" 4 | 5 | 6")
    print("---+---+---")
    print(" 7 | 8 | 9")

    for turn in range(9):

        show_board()

        while True:
            try:
                position = int(input(
                    "Player " + player + ", enter position (1-9): "
                )) - 1

                if position < 0 or position > 8:
                    print("Enter a number from 1 to 9.")
                elif board[position] != " ":
                    print("Position already occupied!")
                else:
                    board[position] = player
                    break

            except ValueError:
                print("Please enter a number.")

        if check_winner(player):
            show_board()
            print("🎉 Player", player, "WINS!")
            return

        if turn == 8:
            show_board()
            print("It's a DRAW!")
            return

        if player == "X":
            player = "O"
        else:
            player = "X"


while True:

    play_game()

    again = input("\nDo you want to play again? (yes/no): ").lower()

    if again != "yes":
        print("Thank you for playing!")
        break
```
