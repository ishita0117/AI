```python
import random

def calculate_cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def hill_climbing(n):
    board = [random.randint(0, n - 1) for _ in range(n)]
    step = 0

    while True:
        step += 1
        current_cost = calculate_cost(board)

        print("Step:", step)
        print("Board:", board)
        print("Cost:", current_cost)

        if current_cost == 0:
            print("Solution Found!")
            break

        best_board = board
        best_cost = current_cost

        for col in range(n):
            for row in range(n):
                if row != board[col]:
                    new_board = board.copy()
                    new_board[col] = row
                    new_cost = calculate_cost(new_board)

                    if new_cost < best_cost:
                        best_cost = new_cost
                        best_board = new_board

        if best_board == board:
            print("Local minimum reached. Restarting...\n")
            board = [random.randint(0, n - 1) for _ in range(n)]
        else:
            board = best_board

    print("Final Board:", board)
    print("Total Steps:", step)
    print("Final Cost:", calculate_cost(board))


n = int(input("Enter number of queens: "))

if n < 1:
    print("Enter a positive number.")
else:
    hill_climbing(n)


```
